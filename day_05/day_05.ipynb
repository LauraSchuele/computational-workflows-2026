### 1. Describe the concept of Workflows, Processes and Channels we deal with in Nextflow

Workflows: A workflow defines how processes are connected through channels. More precise it specifies which processes should run and how the data is transferred between them.





Processes: A process represents an individual computational task. A process includes which inputs are required, the commands to execute and the output we expect. Example: running FastQC on a FASTQ file 






Channels: Stream of data, connecting computational tasks. 
Queue channels: Can emit multiple items. Each item is consumed once by a given subscriber. A Nextflow queue channel does not have to contain all its items before processing begins. Processes can consume items as they become available.
Value channels: Hold a single value that can be reused by multiple process executions.

Task 12 - What do you notice compared to Task 11?

groupTuple() collects all values with the same key into a list, join() matches items from two channels using a common key. A important thing about join() is that it does not support duplicate keys,for example keys 1 and 3 occur twice in the left channel, while key 2 occurs twice in the right channel. Therefore, values such as B, 33 and E may be missing, depending on which duplicates are retained.

Also the output of groupTuple() contains a key and a list of values, whereas join() produces tuples containing the matching values from both channels.
