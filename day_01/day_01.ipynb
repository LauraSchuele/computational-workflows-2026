1. What is nf-core?

nf-core is a community for developing and maintaining nextflow components and pipelines. Its pipelines follow strict guidelines for documentation, testing, versioning and reproducibility, making it easier for researchers to share and reuse computational workflows.

2. How many pipelines are there currently in nf-core?

157 Pipelines can be browesd through on their website. 101 are released, 43 are under development and 13 are archived.

3. Are there any non-bioinformatic pipelines in nf-core?

Yes it seems like there is at least one pipeline called "spinningjenny" which is not used for bioinformatic purposes and simulates spinning yarn. But the vast majority of pipelines seems to be developed for bioinformatic research.

4. Let's go back a couple of steps. What is a pipeline and what do we use it for?

A pipeline automates the steps in typical data processing workflows. The output of one step can become the input of the next step or another subsequent step. The raw input is processed step by step or in parallel in the pipeline to get an informative output. Pipelines also improve reproducibility and reduce manual work.

5. Why do you think nf-core adheres to strict guidelines?

To minimize reproducibility problems and enable collaboration between researchers. The ongoing reproducibility crisis in science is partly caused by using different versions of tools, which can have different dependencies. This can make reproducing another researcher's results difficult. Transparency and strict guidelines could help with this.

6. What are the main features of nf-core pipelines?

Documentation (instructions on how to install, configure and run pipeline), stable releases, open-source code, continuous integration (CI) testing (checks whether changes in the pipeline produce errors), portability (run anywhere, laptops, clusters or cloud platforms), and packaged software dependencies (Tools and their dependencies are packaged using technologies such as Docker to ensure consistent software environments).
