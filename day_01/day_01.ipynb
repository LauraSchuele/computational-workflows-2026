1. What is nf-core?

Nf-core is a community for developing nextflow components and pipelines. It has strict guidelines regarding documentation, stable releases, open source, continuous-integration testing for changes in the pipelines and more.

2. How many pipelines are there currently in nf-core?

157 Pipelines can be browesd through on their website. 101 are released, 43 are under development and 13 are archived.

3. Are there any non-bioinformatic pipelines in nf-core?

Yes it seems like there is at least one pipeline called "spinningjenny" which is not used for bioinformatic purposes. But the vast majority of pipelines seems to be developed for bioinformatic research.

4. Let's go back a couple of steps. What is a pipeline and what do we use it for?

A pipeline automates the steps in typical data processing workflows. The output of one step becomes the input of the next step. The raw input is processed step by step or in parallel in the pipeline to get an informative output.

5. Why do you think nf-core adheres to strict guidelines?

To minimize reproducibility problems and enable collaboration between researchers. The ongoing reproducibility crisis in science is partly caused by using different versions of tools which each have different dependencies, this can make reproducing an result of another researcher hard. Transparency and strict guidelines could help with this.

6. What are the main features of nf-core pipelines?

Documentation, Stable releases, open source, CI testing, run anywhere, packaged software.
