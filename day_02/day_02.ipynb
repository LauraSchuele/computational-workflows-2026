What was the objective of the study?

The objective of the study is to investigate gene expression changes in three different areas 
of mice brains (NAc, mPFC and VTA) with and without chronic pain after oxycodone withdrawl. The researchers aim to answer questions like:
Can chronic pain worsen withdrawal symptoms,considering that Chronic pain and opioid dependence involve overlapping brain reward circuits?
Can the observed molecular changes help identify possible non-opioid medications to help mice with chronic pain during withdrawal?

What do the conditions mean?

oxy: chronic oxycodone injections (for 14 days, then spontaneous withdrawal)


sal: saline injections (control)

What do the genotypes mean?

SNI (= Spared Nerve Injury): mice with prolonged spared nerve injury, they peformed surgery on them to induce chronic pain


Sham: controls, also underwent surgery but no nerves were damaged/ no chronic pain was induced
